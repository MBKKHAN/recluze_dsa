```python
class Point:
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y

    def __str__(self):
        return "[" + str(self.x) + "," + str(self.y) + "]"
```

```python
p1 = Point()
print("p1.x = ", p1.x)
print("p1 ", p1)


p2 = Point(2,4)

print("p2.x = ", p2.x)
print("p2 ", p2)


    
```

```python
class Shape:
    def __init__(self,points):
        self.points = points

    def __str__(self):
        ret = ""
        for i in self.points:
            ret += str(i) + "-"
        return ret[:-1]
        
```

```python
p1 = Point(5,5)
p2 = Point(10,5)
p3 = Point(5,10)

p = [p1,p2,p3]

sh = Shape(p)
```

```python
print(sh)
```

```python
def print_points(self):
    for i in self.points:
        print(str(i))

Shape.print_points = print_points
```

```python
sh.print_points()
```

<h2>Inheritance</h2>

```python
class Triangle(Shape):  #Pass what is in the constructor of Shape while creating object
    pass 
```

```python
t = Triangle(p)
```

```python
t.print_points()
```

```python
def get_area(self):
    vertices = self.points
    n = len(vertices)  # of corners
    a = 0.0
    for i in range(n):
        j = (i + 1) % n
        a += abs(vertices[i].x * vertices[j].y - vertices[j].x * vertices[i].y)
    return a / 2.0

Triangle.get_area = get_area
```

```python
t.get_area()
```
